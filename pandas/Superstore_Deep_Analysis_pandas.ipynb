## 3️⃣ Data Cleaning
Yahan hum missing values, duplicates, aur data types clean karenge.

```python
# Missing values check - column wise
df.isnull().sum()
```

```python
# Missing values percentage
(df.isnull().sum() / len(df) * 100).round(2)
```

```python
# Total missing values pure dataset mein
print("Total missing values:", df.isnull().sum().sum())
print("Koi bhi missing value nahi hai" if df.isnull().sum().sum()==0 else "Missing values maujood hain")
```

```python
# Duplicate rows check
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Agar duplicates hote to yeh unhe drop kar deta (keep first occurrence)
df = df.drop_duplicates(keep='first')
print("Shape after dropping duplicates:", df.shape)
```

```python
# Column names clean karna - spaces ko underscore se replace, lowercase
df.columns = df.columns.str.strip().str.replace(' ', '_').str.replace('-', '_')
df.columns
```

```python
# Order Date aur Ship Date ko proper datetime type mein convert karna
df['Order_Date'] = pd.to_datetime(df['Order_Date'], format='%m/%d/%Y')
df['Ship_Date'] = pd.to_datetime(df['Ship_Date'], format='%m/%d/%Y')
df[['Order_Date', 'Ship_Date']].dtypes
```

```python
# Postal Code ko string mein convert (ID hai, number nahi)
df['Postal_Code'] = df['Postal_Code'].astype('Int64').astype(str)

# Row_ID ko index bhi bana sakte hain, lekin abhi rehne dete hain
df.dtypes
```

```python
# Row ID column hata dena kyunki wo sirf serial number hai, analysis mein useful nahi
df = df.drop(columns=['Row_ID'])
df.head(2)
```

```python
# fillna example - agar koi missing category hoti to hum kaise fill karte
df['Sub_Category'] = df['Sub_Category'].fillna('Unknown')
print("Unique Sub-Categories:", df['Sub_Category'].nunique())
```

```python
# Nayi useful columns derive karna: Order Year, Month, Weekday, aur Shipping Duration (days)
df['Order_Year'] = df['Order_Date'].dt.year
df['Order_Month'] = df['Order_Date'].dt.month
df['Order_Month_Name'] = df['Order_Date'].dt.month_name()
df['Order_Weekday'] = df['Order_Date'].dt.day_name()
df['Shipping_Days'] = (df['Ship_Date'] - df['Order_Date']).dt.days
df[['Order_Date','Ship_Date','Order_Year','Order_Month_Name','Order_Weekday','Shipping_Days']].head()
```

## 7️⃣ GroupBy & Aggregation — Deep Dive

```python
# Category-wise total Sales & Profit
df.groupby('Category')[['Sales','Profit']].sum().round(2)
```

```python
# Region + Category ke combination pe multiple aggregations
df.groupby(['Region','Category']).agg(
    Total_Sales=('Sales','sum'),
    Avg_Sales=('Sales','mean'),
    Total_Profit=('Profit','sum'),
    Orders=('Order_ID','count')
).round(2)
```

```python
# Sub-Category wise average discount, aur profit margin
sub_cat = df.groupby('Sub_Category').agg(
    Total_Sales=('Sales','sum'),
    Total_Profit=('Profit','sum'),
    Avg_Discount=('Discount','mean'),
    Total_Qty=('Quantity','sum')
).round(2)
sub_cat['Profit_Margin_%'] = (sub_cat['Total_Profit'] / sub_cat['Total_Sales'] * 100).round(2)
sub_cat.sort_values('Total_Profit', ascending=False)
```

```python
# groupby().transform() - har row mein uske Category ka average Sales add karna
df['Category_Avg_Sales'] = df.groupby('Category')['Sales'].transform('mean').round(2)
df[['Category','Sales','Category_Avg_Sales']].head()
```

```python
# groupby().filter() - sirf wo States rakhna jinka total sales 50,000 se zyada hai
state_sales = df.groupby('State').filter(lambda x: x['Sales'].sum() > 50000)
print("States with total sales > 50000 -> total rows:", state_sales.shape[0])
print("Number of qualifying states:", state_sales['State'].nunique())
```

```python
# Multiple agg functions ek saath (min, max, mean, count, std)
df.groupby('Segment')['Profit'].agg(['min','max','mean','std','count']).round(2)
```

## 8️⃣ Pivot Table & Cross-tabulation

```python
# Pivot table - Region (rows) vs Category (columns) -> Total Sales
pivot1 = pd.pivot_table(df, values='Sales', index='Region', columns='Category', aggfunc='sum', margins=True).round(2)
pivot1
```

```python
# Pivot table - Segment vs Ship Mode -> Average Profit
pivot2 = pd.pivot_table(df, values='Profit', index='Segment', columns='Ship_Mode', aggfunc='mean').round(2)
pivot2
```

## 9️⃣ Reshaping Data — melt, stack, unstack

```python
# pivot_table ka result stack/unstack karna
region_cat_sales = df.pivot_table(values='Sales', index='Region', columns='Category', aggfunc='sum')
stacked = region_cat_sales.stack()
print("Stacked (long format):")
stacked.head(8)
```

```python
# unstack - wapas wide format mein
unstacked = stacked.unstack()
unstacked.round(2)
```

```python
# melt - wide format ko long format mein convert karna
sample_wide = df.groupby('Region')[['Sales','Profit','Quantity']].sum().reset_index()
melted = pd.melt(sample_wide, id_vars='Region', value_vars=['Sales','Profit','Quantity'],
                  var_name='Metric', value_name='Value')
melted
```

## 🔟 apply(), map(), str & dt Accessors

```python
# apply() with lambda - profit ko categorize karna (Loss / Low / High)
def profit_category(p):
    if p < 0:
        return 'Loss'
    elif p < 50:
        return 'Low Profit'
    else:
        return 'High Profit'

df['Profit_Category'] = df['Profit'].apply(profit_category)
df['Profit_Category'].value_counts()
```

```python
# map() - Segment ko short codes mein map karna
segment_map = {'Consumer': 'CON', 'Corporate': 'CORP', 'Home Office': 'HO'}
df['Segment_Code'] = df['Segment'].map(segment_map)
df[['Segment','Segment_Code']].drop_duplicates()
```

```python
# str accessor - Customer Name se First Name nikalna
df['Customer_First_Name'] = df['Customer_Name'].str.split(' ').str[0]
df[['Customer_Name','Customer_First_Name']].head()
```

```python
# str accessor - Product Name uppercase/lowercase, aur length
df['Product_Name_Length'] = df['Product_Name'].str.len()
df[['Product_Name','Product_Name_Length']].sort_values('Product_Name_Length', ascending=False).head(3)
```

```python
# dt accessor - quarter aur is_month_end jaise attributes
df['Order_Quarter'] = df['Order_Date'].dt.quarter
df['Is_Weekend_Order'] = df['Order_Date'].dt.dayofweek.isin([5,6])
df[['Order_Date','Order_Quarter','Order_Weekday','Is_Weekend_Order']].head()
```

```python
# applymap ka chota example - numeric columns ko round karna
numeric_cols = ['Sales','Profit','Discount']
df[numeric_cols] = df[numeric_cols].apply(lambda col: col.round(2))
df[numeric_cols].head()
```

## 1️⃣1️⃣ Binning — cut() & qcut()

```python
# cut() - Sales ko fixed-width bins mein divide karna
df['Sales_Bin'] = pd.cut(df['Sales'], bins=[0,50,200,500,1000,df['Sales'].max()],
                           labels=['Very Low','Low','Medium','High','Very High'])
df['Sales_Bin'].value_counts()
```

```python
# qcut() - Profit ko quartiles (equal-sized bins) mein divide karna
df['Profit_Quartile'] = pd.qcut(df['Profit'], q=4, labels=['Q1(Lowest)','Q2','Q3','Q4(Highest)'])
df['Profit_Quartile'].value_counts()
```

## 1️⃣2️⃣ Correlation, Ranking & Cumulative Stats

```python
# Correlation matrix - numeric columns ke beech relationship
corr_matrix = df[['Sales','Quantity','Discount','Profit']].corr().round(3)
corr_matrix
```

```python
# rank() - Sub-Category ko total profit ke hisaab se rank dena
sub_profit = df.groupby('Sub_Category')['Profit'].sum().sort_values(ascending=False)
ranked = sub_profit.rank(ascending=False).astype(int)
pd.DataFrame({'Total_Profit': sub_profit.round(2), 'Rank': ranked})
```

```python
# cumsum() - date order se sorted sales ka cumulative sum (running total)
daily_sales = df.groupby('Order_Date')['Sales'].sum().sort_index()
cumulative_sales = daily_sales.cumsum()
print("Total cumulative sales (end):", round(cumulative_sales.iloc[-1], 2))
cumulative_sales.head()
```

```python
# cummax / cummin example
print("Running max daily sales (last 5):")
print(daily_sales.cummax().tail())
```

## 1️⃣3️⃣ Time Series Analysis — resample() & rolling()

```python
# Order_Date ko index bana ke month-wise resample karna
ts = df.set_index('Order_Date').sort_index()
monthly_sales = ts['Sales'].resample('ME').sum()
monthly_sales.head(12)
```

```python
# resample - quarter-wise total profit
quarterly_profit = ts['Profit'].resample('QE').sum().round(2)
quarterly_profit
```

```python
# rolling() - 3-month rolling average of monthly sales (trend smoothing)
rolling_avg = monthly_sales.rolling(window=3).mean().round(2)
pd.DataFrame({'Monthly_Sales': monthly_sales.round(2), '3M_Rolling_Avg': rolling_avg}).head(8)
```

## 1️⃣4️⃣ idxmax/idxmin & Other Handy Functions

```python
# Sabse zyada profit wala Sub-Category dhoondna
sub_cat_profit = df.groupby('Sub_Category')['Profit'].sum()
print("Highest profit Sub-Category:", sub_cat_profit.idxmax(), "->", round(sub_cat_profit.max(),2))
print("Lowest (most loss) Sub-Category:", sub_cat_profit.idxmin(), "->", round(sub_cat_profit.min(),2))
```

```python
# Sabse zyada orders dene wala Customer
top_customer = df['Customer_Name'].value_counts().idxmax()
print("Most frequent customer:", top_customer, "with", df['Customer_Name'].value_counts().max(), "orders")
```

```python
# Sabse zyada revenue dene wala State
state_revenue = df.groupby('State')['Sales'].sum().sort_values(ascending=False)
state_revenue.head(10)
```

## 1️⃣5️⃣ merge() & concat() Demo

```python
# Do chhote summary DataFrames banate hain aur unhe merge karte hain
region_sales = df.groupby('Region')['Sales'].sum().reset_index().rename(columns={'Sales':'Total_Sales'})
region_profit = df.groupby('Region')['Profit'].sum().reset_index().rename(columns={'Profit':'Total_Profit'})

# merge - dono ko Region column pe join karna
region_summary = pd.merge(region_sales, region_profit, on='Region', how='inner')
region_summary['Profit_Margin_%'] = (region_summary['Total_Profit'] / region_summary['Total_Sales'] * 100).round(2)
region_summary
```

```python
# concat - do saal (2016, 2017) ke data ko alag-alag nikal ke phir concat karna
data_2016 = df[df['Order_Year'] == 2016]
data_2017 = df[df['Order_Year'] == 2017]
combined = pd.concat([data_2016, data_2017], axis=0)
print("2016 rows:", len(data_2016), "| 2017 rows:", len(data_2017), "| Combined:", len(combined))
```
